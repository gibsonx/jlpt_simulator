## Section 4. Listening Comprehension

### 問題1 では、まず質問を聞いてください。それから話を聞いて、問題用紙の1から4の中から、最もよいものを一つえらんでください。 (topic_understanding)

```python
graph = GraphBuilder(exam_uid=exam_uid)
factory = JLPTTaskFactory(graph, level=level)
content = factory.topic_understanding_txt(random.choice(topics),seq=1)
display(HTML(_render_result(result=content)))
```

### 問題2 では、まず質問を聞いてください。そのあと、問題用紙を見てください。読む時間があります。それから話を聞いて、問題用紙の1から4の中から、最もよいものを一つえらんでください。
 (keypoint_understanding)

```python
graph = GraphBuilder(exam_uid=exam_uid)
factory = JLPTTaskFactory(graph, level=level)
content = factory.keypoint_understanding(random.choice(topics),seq=2)
display(HTML(_render_result(result=content)))
```

### 問題3では、問題用紙（もんだいようし）に何（なに）も いんさつされていません。この問題（もんだい）は、ぜんたいとして どんな ないようかを聞（き）く 問題（もんだい）です。話（はなし）の前（まえ）に 質問（しつもん）は ありません。まず 話（はなし）を 聞（き）いてください。それから、質問（しつもん）と せんたくし を聞（き）いて、1から4の中（なか）から、最（もっと）も よい ものを 一（ひと）つ えらんでください。

```python
graph = GraphBuilder(exam_uid=exam_uid)
factory = JLPTTaskFactory(graph, level=level)
content = factory.summary_understanding(random.choice(topics),seq=2)
display(HTML(_render_result(result=content)))
### 問題4 では、元を見ながら質問を聞いてください。やじるし（➔）の人は何と言いますか。1 から 3 の中から、最もよいものを一つえらんでください。 (active_expression)
```

```python
graph = GraphBuilder(exam_uid=exam_uid)
factory = JLPTTaskFactory(graph, level=level)
content = factory.active_expression((random.choice(topics)),seq=3)
display(HTML(_render_result(result=content)))
```

### 問題5 では、問題用紙に何もいんさつしていません。まず文を聞いてください。それから、そのへんしを聞いて、1 から 3 の中から、最もよいものを一つえらんでください。 (immediate_ack)

```python
graph = GraphBuilder(exam_uid=exam_uid)
factory = JLPTTaskFactory(graph, level=level)
content = factory.immediate_ack((random.choice(topics)),seq=4)
display(HTML(_render_result(result=content)))
```
